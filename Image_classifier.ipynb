### 🌌 Custom Edition: Cosmic Cartographer & Galactic Specimen Analyzer
This specialized version features a celestial-nebula styling pipeline, dynamic space-classification heuristics, and a futuristic sci-fi laboratory HUD.

```python
import gradio as gr
import torch
import os
from PIL import Image, ImageOps
from transformers import BlipProcessor, BlipForConditionalGeneration

# ==========================================================
# 1. OPTIMIZED ADVANCED MODEL LOADING
# ==========================================================

device = "cuda" if torch.cuda.is_available() else "cpu"

# Keeping the high-performance BLIP-Large Model initialized
if "upgraded_processor" not in globals() or "upgraded_model" not in globals():
    print("Initializing Advanced BLIP-Large Model...")
    upgraded_processor = BlipProcessor.from_pretrained(
        "Salesforce/blip-image-captioning-large"
    )
    upgraded_model = BlipForConditionalGeneration.from_pretrained(
        "Salesforce/blip-image-captioning-large"
    ).to(device)
    upgraded_model.eval()

print("Scan Core Loaded Successfully!")

# ==========================================================
# 2. CORE INTELLIGENCE: LIGHTWEIGHT CLASSIFIER
# ==========================================================

def analyze_specimen(image, scan_mode):
    if image is None:
        return None, "Please provide an image.", None, "Ready."

    try:
        # Ensure correct orientation
        image = ImageOps.exif_transpose(image).convert("RGB")

        # Generate High-Fidelity Description
        inputs = upgraded_processor(images=image, return_tensors="pt").to(device)
        with torch.inference_mode():
            outputs = upgraded_model.generate(
                **inputs,
                max_new_tokens=80,
                num_beams=3,
                repetition_penalty=1.2
            )

        raw_caption = upgraded_processor.decode(outputs[0], skip_special_tokens=True)
        caption = raw_caption.capitalize()

        # Dynamic Science Classifications
        lower_caption = caption.lower()

        # Keyword Matcher
        if any(kw in lower_caption for kw in ["cat", "dog", "bird", "animal", "fur", "wild", "beast", "horse", "fish", "human", "person"]):
            specimen_class = "Bio-Organic (Fauna/Flora)"
            risk_level = "Standard"
            composition = "Carbon-Based Biological Structure"
        elif any(kw in lower_caption for kw in ["computer", "phone", "screen", "laptop", "keyboard", "device", "monitor", "wire", "gadget", "metal", "robot"]):
            specimen_class = "Technology / Electronics"
            risk_level = "Inert"
            composition = "Silicon & Metallic Components"
        elif any(kw in lower_caption for kw in ["tree", "flower", "grass", "mountain", "sky", "river", "lake", "outdoor", "forest", "plant"]):
            specimen_class = "Natural Landscape / Vegetation"
            risk_level = "Safe"
            composition = "Organic Plant Matter / Mineral"
        elif any(kw in lower_caption for kw in ["car", "bus", "truck", "bike", "motorcycle", "wheel", "train", "vehicle"]):
            specimen_class = "Mechanical Transport"
            risk_level = "Controlled"
            composition = "Metal Alloy & Rubber"
        else:
            specimen_class = "Unidentified Galactic Artifact"
            risk_level = "Uncharted"
            composition = "Crystalline / Dark Matter Catalyst"

        # Dynamic Match Score
        if scan_mode == "Deep Scan":
            match_score = 98
        else:
            match_score = min(45 + len(caption) % 40, 95)

        # Cyberpunk Neon-Dark Theme HTML Presentation Card
        html_output = f"""
        <div class='cosmic-card'>
            <div class='card-header-cosmic'>
                <span class='cosmic-badge'>SCAN ACTIVE</span>
                <span class='signal-strength'>COGNITION MATCH: {match_score}%</span>
            </div>
            <h2 class='cosmic-title'>🌌 GALACTIC MANIFEST REPORT</h2>

            <div class='scan-description'>
                <p class='desc-label'>NEURAL TRANSLATED CAPTION</p>
                <p class='desc-text'>"{caption}"</p>
            </div>

            <div class='stats-grid'>
                <div class='stat-box'>
                    <span class='stat-lbl'>SPECIMEN CLASS</span>
                    <span class='stat-val highlight-neon-blue'>{specimen_class}</span>
                </div>
                <div class='stat-box'>
                    <span class='stat-lbl'>HAZARD LEVEL</span>
                    <span class='stat-val highlight-neon-red'>{risk_level}</span>
                </div>
                <div class='stat-box'>
                    <span class='stat-lbl'>MOLECULAR COMPOSITION</span>
                    <span class='stat-val'>{composition}</span>
                </div>
                <div class='stat-box'>
                    <span class='stat-lbl'>SCANNING PROTOCOL</span>
                    <span class='stat-val'>{scan_mode}</span>
                </div>
            </div>
        </div>
        """

        # Export Manifest file
        report_path = "analysis_manifest.txt"
        with open(report_path, "w") as f:
            f.write("=== COSMIC SPECIMEN REPORT ===\n")
            f.write(f"Confidence: {match_score}%\n")
            f.write(f"Assigned Classification: {specimen_class}\n")
            f.write(f"Composition: {composition}\n")
            f.write(f"Hazard Level: {risk_level}\n")
            f.write(f"Caption: {caption}\n")
            f.write("==============================\n")

        return image, html_output, report_path, "Cosmic Scan completed successfully."

    except Exception as e:
        error_html = f"<div style='color: #ff3366; padding: 15px;'><b>Error:</b> {str(e)}</div>"
        return None, error_html, None, "Scan failed."

# ==========================================================
# 3. FUTURISTIC NEON STYLE SPECIFICATION
# ==========================================================

cosmic_css = """
.gradio-container {
    max-width: 1200px !important;
    margin: auto !important;
    background-color: #0b0f19 !important;
    font-family: 'Courier New', Courier, monospace !important;
}

#cosmic-banner {
    background: linear-gradient(135deg, #111827, #030712);
    border: 1px solid #00f2fe;
    box-shadow: 0 0 15px rgba(0, 242, 254, 0.3);
    padding: 30px;
    border-radius: 12px;
    margin-bottom: 25px;
    text-align: center;
}

#cosmic-banner h1 {
    font-size: 32px;
    font-weight: 900;
    color: #00f2fe;
    text-shadow: 0 0 10px rgba(0, 242, 254, 0.8);
    margin: 0 0 8px 0;
    letter-spacing: 2px;
}

#cosmic-banner p {
    color: #9ca3af;
    font-size: 14px;
    margin: 0;
}

.cosmic-panel {
    background: #0f172a !important;
    border: 1px solid #1e293b !important;
    border-radius: 12px !important;
    padding: 20px !important;
    box-shadow: inset 0 0 10px rgba(0, 0, 0, 0.5);
}

#action-btn {
    background: linear-gradient(90deg, #00f2fe, #4facfe) !important;
    color: #000000 !important;
    border: none !important;
    font-weight: 900 !important;
    border-radius: 8px !important;
    box-shadow: 0 0 10px rgba(0, 242, 254, 0.5) !important;
    transition: all 0.2s ease;
    letter-spacing: 1px;
}

#action-btn:hover {
    box-shadow: 0 0 20px rgba(0, 242, 254, 0.9) !important;
    transform: scale(1.02);
}

.cosmic-card {
    padding: 10px;
    border: 1px dashed #38bdf8;
    background: rgba(15, 23, 42, 0.8);
    border-radius: 10px;
}

.card-header-cosmic {
    display: flex;
    justify-content: space-between;
    align-items: center;
    margin-bottom: 12px;
}

.cosmic-badge {
    background: #00f2fe;
    color: #000000;
    font-weight: 800;
    padding: 3px 8px;
    border-radius: 4px;
    font-size: 11px;
    box-shadow: 0 0 5px rgba(0, 242, 254, 0.8);
}

.signal-strength {
    color: #38bdf8;
    text-shadow: 0 0 5px rgba(56, 189, 248, 0.5);
    font-weight: 700;
    font-size: 14px;
}

.cosmic-title {
    color: #e2e8f0;
    font-size: 20px;
    font-weight: 800;
    margin-bottom: 15px;
    border-bottom: 1px solid #1e293b;
    padding-bottom: 8px;
}

.scan-description {
    background: rgba(30, 41, 59, 0.5) !important;
    border: 1px solid #38bdf8;
    padding: 15px;
    border-radius: 8px;
    margin-bottom: 15px;
}

.desc-label {
    font-size: 11px;
    color: #38bdf8;
    font-weight: 700;
    margin: 0 0 5px 0;
    letter-spacing: 1px;
}

.desc-text {
    font-size: 14px;
    color: #f1f5f9;
    margin: 0;
    font-style: italic;
}

.stats-grid {
    display: grid;
    grid-template-columns: 1fr 1fr;
    gap: 12px;
}

.stat-box {
    background: #090d16;
    padding: 12px;
    border-radius: 8px;
    border: 1px solid #1e293b;
}

.stat-lbl {
    display: block;
    font-size: 10px;
    color: #94a3b8;
    font-weight: 700;
    margin-bottom: 5px;
    letter-spacing: 0.5px;
}

.stat-val {
    font-size: 13px;
    color: #f1f5f9;
    font-weight: 600;
}

.highlight-neon-blue {
    color: #00f2fe;
    text-shadow: 0 0 5px rgba(0, 242, 254, 0.4);
}

.highlight-neon-red {
    color: #ff3366;
    text-shadow: 0 0 5px rgba(255, 51, 102, 0.4);
}
"""

# ==========================================================
# 4. INTERFACE ASSEMBLY
# ==========================================================

with gr.Blocks(title="Cosmic Cartographer Studio") as clean_demo:

    # Header Element
    gr.HTML("""
    <div id="cosmic-banner">
        <h1>🌌 COSMIC CARTOGRAPHER</h1>
        <p>A sci-fi laboratory HUD for visual telemetry analysis, planetary specimen classification, and stellar diagnostics.</p>
    </div>
    """)

    with gr.Row():
        # Control Deck
        with gr.Column(scale=5, elem_classes=["cosmic-panel"]):
            gr.Markdown("### 🛰️ TELEMETRY CONSOLE")

            img_input = gr.Image(
                type="pil",
                label="Target Image Scan Specimen",
                height=300
            )

            scan_mode = gr.Radio(
                choices=["Quick Scan", "Deep Scan"],
                value="Quick Scan",
                label="Core Analysis Resolution"
            )

            analyze_btn = gr.Button(
                "⚡ INITIATE DENSE SCAN",
                elem_id="action-btn"
            )

            clear_btn = gr.ClearButton(
                components=[img_input],
                value="RESET CORE"
            )

        # Diagnostic Lab
        with gr.Column(scale=6, elem_classes=["cosmic-panel"]):
            gr.Markdown("### 📊 SPECTRAL DIAGNOSTICS")

            img_preview = gr.Image(
                label="Optical Signal Feed",
                height=200,
                interactive=False
            )

            analysis_html = gr.HTML(
                value="""
                <div class='cosmic-card'>
                    <div class='card-header-cosmic'>
                        <span class='cosmic-badge' style='background: #ff3366; color: #000;'>STANDBY</span>
                    </div>
                    <h2 class='cosmic-title'>No Target Loaded</h2>
                    <div class='scan-description'>
                        <p class='desc-label'>SYSTEM STATUS</p>
                        <p class='desc-text'>Please load a visual target into the scanner to initialize telemetry capture.</p>
                    </div>
                </div>
                """
            )

            with gr.Row():
                report_file = gr.File(
                    label="📥 Extract Data Manifest",
                    interactive=False
                )
                status_box = gr.Textbox(
                    label="Quantum Stream Feed",
                    value="Ready for Input",
                    interactive=False
                )

    # Action Configuration
    analyze_btn.click(
        fn=analyze_specimen,
        inputs=[img_input, scan_mode],
        outputs=[img_preview, analysis_html, report_file, status_box]
    )

# ==========================================================
# 5. EXECUTE INTERFACE (Fixed Gradio 6.0 Parameters)
# ==========================================================
clean_demo.queue()
clean_demo.launch(theme=gr.themes.Default(), css=cosmic_css, share=True, show_error=True)
```
